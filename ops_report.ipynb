```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/katiegaertner/bramble-ops-report/refs/heads/main/cafe_log.csv')
```

```python
def waste_rate(data):
    total_waste_lbs = data['waste_lbs'].sum()
    total_roasted_lbs = data['roasted_lbs'].sum()
    return total_waste_lbs / total_roasted_lbs
```

```python
march_df = df[df['month'] == 'March']
print(f"March Waste Rate: {waste_rate(march_df):.1%}")
```

```python
march_cafes = march_df['cafe_id'].nunique()
print(f"March summary: {waste_rate(march_df):.1%} waste across {march_cafes} cafes.")
```

```python
april_df = df[df['month'] == 'April']
print(f"April Waste Rate: {waste_rate(april_df):.1%}")
```

```python
april_cafes = april_df['cafe_id'].nunique()
print(f"April summary: {waste_rate(april_df):.1%} waste across {april_cafes} cafes.")
```
